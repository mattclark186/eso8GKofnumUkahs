# Lazy Predict

This notebook uses lazy predict, a package that quickly builds many different types of models using default parameters and displays various metrics from them. This is used to get understand which models may be good candidates and require further investigation.


The outcome is that the models worth looking in to further are:
- KNeighbors
- BernoulliNB
- RandomForest
- XGBoost

```python
# import all the necessary libraries
import numpy as np
import pandas as pd
from lazypredict.Supervised import LazyClassifier
from sklearn.model_selection import train_test_split

from src.data.prepare_data import x_y_split
```

```python
# import the data and split it into x and y (features and target)
data = pd.read_csv('data/raw/ACME-HappinessSurvey2020.csv')
x, y = x_y_split(data)
```

```python
# splitting the data into training and testing
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=29, stratify=y)

# building and fitting the classifier
lzclf = LazyClassifier(verbose=0, ignore_warnings=True, custom_metric=None)
models, predictions = lzclf.fit(x_train, x_test, y_train, y_test)

# display the results
models
```
