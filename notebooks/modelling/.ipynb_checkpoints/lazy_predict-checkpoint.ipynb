```python
import numpy as np
import pandas as pd
from lazypredict.Supervised import LazyClassifier
from sklearn.metrics import f1_score

from src.data.prepare_data import x_y_split, tt_split
```

```python
data = pd.read_csv('data/raw/ACME-HappinessSurvey2020.csv')

x, y = x_y_split(data)

x_train, x_test, y_train, y_test = tt_split(x, y)
```

```python
def custom_f1(y_true, y_pred):
    return f1_score(y_true, y_pred, average='weighted')

lzclf = LazyClassifier(verbose=0, ignore_warnings=True, custom_metric=custom_f1)
models, predictions = lzclf.fit(x_train, x_test, y_train, y_test)
```

```python
from sklearn.model_selection import train_test_split

data = pd.read_csv('data/raw/ACME-HappinessSurvey2020.csv')

x = data.drop(columns=['Y'])
y = data['Y']

x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=1)

lzclf = LazyClassifier(verbose=0, ignore_warnings=True, custom_metric=custom_f1)
models, predictions = lzclf.fit(x_train, x_test, y_train, y_test)

models.loc[['XGBClassifier', 'RandomForestClassifier', 'KNeighborsClassifier']]
```
